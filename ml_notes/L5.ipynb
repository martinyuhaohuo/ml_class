# Neural Networks

## Shallow Neural Networks

### Single Input and Output Neural network
- Consider a simple neural network: 
    $$
    \begin{align*}
    \displaystyle f[x,\phi] &= \phi_0 + \phi_1 \underbrace{a(\theta_{10}+\theta_{11}x)}_{h_1} + \phi_2 \underset{h_2}{\underbrace{a(\theta_{20}+\theta_{21}x)}} + \phi_3 \underset{h_3}{\underbrace{a(\theta_{30}+\theta_{31}x)}} \\ a(z) &= ReLU(z) = \max\{z,0\} 
    \end{align*}
    $$
- We can break down this neural network into three parts: 
    - First, we compute three linear functions of the input data ($θ_{10} + θ_{11}x$, $θ_{20} + θ_{21}x$, and $θ_{30} + θ_{31}x$)
    - Second, we pass the three results through an **activation function** $a[•]$
    - Finally, we weight the three resulting activations with $ϕ_1, ϕ_2, ϕ_3$, sum them, and add an offset (also called bias) $ϕ_0$

### Activation Function
- An activation function is a non-linear function:
    - There are many possibilities, but the most common choice is the rectified linear unit function (ReLU)
    - ReLU: $a[z] = \mathrm{ReLU}[z] =\begin{cases} 0 & z < 0 \\ z & z \ge 0 \end{cases}$
- Activation pattern of ReLU:
    - Note that the linear function output is clipped by the ReLU activation function for values below zero
    - When a unit is clipped, we refer to it as inactive
    - When it is not clipped, we refer to it as active

### Hidden Units
- We refer to $h_1$, $h_2$, and $h_3$ as hidden units
- Input of hidden unit:
    - The input of each hidden unit is the univariate input x
- Pre-activation:
    - Each hidden unit first apply a linear function on input x
    - This step is called **pre-activation**
- Activation:
    - Then, the output of the linear function is inputted into a non-linear activation fuynction (such as ReLU) 
    - This step is called **activation** and forms the output of each hidden unit
- Aggregate output of hidden units:
    - In a shallow neural network, there is only one layer of hidden units
    - Therefore, the network applyies a linear combintaion on the outputs of hidden units plus a bias term to form the final prediction
- Intuition:
    - How neural network add felxibility (also called network capacity) to output function?
    - Each hidden unit contributes one “joint” to the output function
    - So with three hidden units, there can be four linear regions in the output function
    - However, only three of the slopes of these regions are independent
    - The fourth is either zero (if all the hidden units are inactive in this region) or is a sum of slopes from the other regions

### Universal Approximation Theorem
- Consider a shallow neural network with one input, one output, and D hidden units
    - The $d$ th hidden unit is defined: $h_d = a\!\left[\theta_{d0} + \theta_{d1} x \right]$
    - The model is defined as: $y = \phi_0 + \sum_{d=1}^{D} \phi_d h_d$
- With ReLU activation functions, the output of a network with D hidden units has at most D joints 
    - It is a piecewise linear function with at most D + 1 linear regions
    - As we add more hidden units, the model can approximate more complex functions
- The universal approximation theorem proves that for any continuous function (does not limit to 1D function, but also function with multi-nomial input and output), there exists a shallow network that can approximate this function to any specified precision

### Multivariate Input and Output Neural Network
- Multivariate output:
    -  A network with a scalar input $x$, four hidden units $h_1$, $h_2$, $h_3$, and $h_4$, and a 2D multivariate output $y = [y_1, y_2]'$ would be defined as:
        - $h_1 = a[\theta_{10} + \theta_{11} x]$
        - $h_2 = a[\theta_{20} + \theta_{21} x]$
        - $h_3 = a[\theta_{30} + \theta_{31} x]$
        - $h_4 = a[\theta_{40} + \theta_{41} x]$
        - $y_1 = \phi_{10} + \phi_{11} h_1 + \phi_{12} h_2 + \phi_{13} h_3 + \phi_{14} h_4$
        - $y_2 = \phi_{20} + \phi_{21} h_1 + \phi_{22} h_2 + \phi_{23} h_3 + \phi_{24} h_4$
    - The two outputs are two different linear functions of the hidden units
    - The “joints” in the output functions depend on where the linear functions in hidden units $θ_0 + θ_1x$ are clipped by the ReLU functions
    - Since both outputs y1 and y2 are different linear functions of the same four hidden units, the “joints” in each must be in the same places
    - The slopes of the linear regions and the overall vertical offset can differ between y1 and y2
- Multivariate input:
    - A network with two inputs $x = [x_1, x_2]'$ and a scalar output y with three hidden units would be defined as:
        - $h_1 = a[\theta_{10} + \theta_{11} x_1 + \theta_{12} x_2]$
        - $h_2 = a[\theta_{20} + \theta_{21} x_1 + \theta_{22} x_2]$
        - $h_3 = a[\theta_{30} + \theta_{31} x_1 + \theta_{32} x_2]$
        - $y = \phi_0 + \phi_1 h_1 + \phi_2 h_2 + \phi_3 h_3$
    - The hidden units are combined to form the output in the usual way
- Multivariate input and output shallow network (general case):
    - We now define a general equation for a shallow neural network $\mathbf{y} = f[\mathbf{x}, \boldsymbol{\phi}]$
    - It maps a multi-dimensional input $\mathbf{x} \in \mathbb{R}^{D_i}$ to a multi-dimensional output $\mathbf{y} \in \mathbb{R}^{D_o}$ using hidden units $\mathbf{h} \in \mathbb{R}^{D}$
    - Each hidden unit is defined as 
        $$ h_d = a\!\left[\theta_{d0} + \sum_{i=1}^{D_i} \theta_{di} x_i\right]$$
        - Where $a[•]$ is a nonlinear activation function
        - The activation function permits the model to describe nonlinear relations between input and the output
        - It must be nonlinear itself, with a linear activation function, the overall mapping from input to output would be restricted to be linear
    - The j-th element in the output vector $\mathbf{y}$ is $$y_j = \phi_{j0} + \sum_{d=1}^{D} \phi_{jd} h_d$$

### Some Terminology
- Input layer: the input vector
- Hidden layer: ensembles of hidden units
- Output layer: a linear combination of hidden units' outputs plus a bias term
- Neurons: hidden unit
- Pre-activation: values after applying linear function on the inputs to the hidden layer
- Activation: the values at the hidden layer after the ReLU functions
- Multi-layer perceptron: any neural network with at least one hidden layer
- Shallow neural network: networks with one hidden layer
- Deep neural network: networks with multiple hidden layers
- Weights: slope parameters in the underlying equations
- Biases: offset parameters in the underlying equations

## Deep Neural Networks

### Composing Shallow Networks (Special Case)
- Composing shallow network is a special case of deep neural network
    - We use composing shallow network to gain insight into the behavior of deep neural networks
    - Composing two shallow networks is equivalent to a deep network with two hidden layers
- Consider composing two shallow networks (univariate input anf output) each with three hidden units:
    - The output of the first becomes the input of the second
    - With ReLU activations, this model's output function also describes a family of piecewise linear functions
    - However, the number of linear regions ("number of joint") in the output function is potentially greater than for a shallow network with six hidden units
    - More specifically, this composing shallow network with six hidden units has 8 joints, while a shallow network with six hidden units only have 6 joints
- This lead to the reason for why we need deep network:
    - Deep network gains deep efficiency
    - With same number of parameters (e.g. hidden units), the deep network results in more linear regions than shallow network

### Deep Neural Network General Case
- Modern deep networks might have more than a hundred layers with thousands of hidden units at each layer
- The number of hidden units in each layer is referred to as the **width** of the network, denoted as $D_k$
- The number of hidden layers  is referred to as the **depth**, denoted as $K$
- The total number of hidden units is a measure of the network’s **capacity**, denoted as $\sum_{k = 1}^{K}D_k$
- The input vector $x$ and output vector $y$ is of any dimension
- The vector of bias is $\beta_k$, with dimension $D_{k-1}$ ($D_{0}$ is dimension of input vector, $D_{K+1}$ is dimension of output vector)
- The matrix of weight is $\Omega_k$, with dimension $D_{k} \times D_{k-1}$
- The hidden layers $h_k$ is a vector of $D_k$ dimension
- $a[•]$ is the element-wise activation function
$$\begin{align*}
h_1 & = a[\beta_1 + \Omega_1 x] \\
h_2 & = a[\beta_2 + \Omega_2 h_1] \\
h_3 & = a[\beta_3 + \Omega_3 h_2] \\
& \vdots \\
h_K & = a[\beta_{K} + \Omega_{K} h_{K-1}] \\
y & = \beta_{K+1} + \Omega_{K+1} h_K
\end{align*}
$$

- The below figure shows how maximum number of linear regions increases as a function of the number of parameters for univariate-input-output networks
- With same parameter number, deep neural networks (large number of hidden layers) create more complex functions (more linear regions) than shallow neural networks
- This effect would be magnified as the number of input dimensions increases
- Depth efficiency of neural networks:
    - Some functions have been identified that require a shallow network with exponentially more hidden units to achieve an equivalent approximation to that of a deep network

## Backpropagation

### Introduction to the Problem
- To update parameters of a neural network, we need to calculate the derivative with respect to each parameter
- However, for a deep neural network, there could be million parameters, and the derivative needs to be computed for every iteration of the training
- This makes how to calculate the derivatives efficiently a central problem in optimization algorithm
- **Backpropagation algorithm** is a method to compute gradients efficiently

### Set Up of the Problem
- Consider a network $f[x, ϕ]$ with multivariate input $x$, parameters $ϕ$, and three hidden layers $h_1$, $h_2$, and $h_3$
$$
\begin{align*}
\mathbf{h}_1 &= \mathbf{a}\left[\boldsymbol{\beta}_0 + \boldsymbol{\Omega}_0 \mathbf{x}\right] \\
\mathbf{h}_2 &= \mathbf{a}\left[\boldsymbol{\beta}_1 + \boldsymbol{\Omega}_1 \mathbf{h}_1\right] \\
\mathbf{h}_3 &= \mathbf{a}\left[\boldsymbol{\beta}_2 + \boldsymbol{\Omega}_2 \mathbf{h}_2\right] \\
f[\mathbf{x}, \boldsymbol{\phi}] &= \boldsymbol{\beta}_3 + \boldsymbol{\Omega}_3 \mathbf{h}_3
\end{align*}
$$
- The activation function $a[.]$ is ReLU function
- The model parameters $ϕ = \{β_0, Ω_0, β_1, Ω_1, β_2, Ω_2, β_3, Ω_3 \}$ consist of the bias vectors $β_k$ and weight matrices $Ω_k$ of every layer
- We also have individual loss terms $ℓ_i$, the stochastic gradient descent (SGD) algorithm will update parameter as: 
$$
\boldsymbol{\phi}_{t+1} \leftarrow \boldsymbol{\phi}_t
- \alpha \sum_{i \in \mathcal{B}_t}
\frac{\partial \ell_i[\boldsymbol{\phi}_t]}{\partial \boldsymbol{\phi}}
$$
- To compute this update, we need to calculate the derivatives $\frac{\partial \ell_i}{\partial \boldsymbol{\beta}_k}$ and $\frac{\partial \ell_i}{\partial \boldsymbol{\Omega}_k}$ at every layer $k ∈ {0, 1, . . . , K}$ and for every $i$ belong to the batch 
- That is, we need to compute the derivative of individual loss w.r.t each parameter at each layer (hidden or output layer)

### Idea of Backward Propogation
- Observation 1: 
    - Each weight (element of $Ω_k$) multiplies the activation at a source hidden unit
    - Hence, the effect of any change to the weight (derivative w.r.t each weight) would depend on the value of the activation
    - That is, when calculating the derivative w.r.t each weight, we will need the value of actrivtion of each hidden unit
    - To prevent repeative compution, we can run the network for each data point in the batch and store the activations of all the hidden units
    - This is known as the **forward pass**, and the stored activations will subsequently be used to compute the gradients
- Observation 2:
    - A small change in a bias or weight causes a chain effect of changes through the subsequent layers
    - Hence, to know how changing a parameter modifies loss, we also need to know how changes to every subsequent hidden layer will modify their successor
    - For example, to calculate how a small change in a weight/bias in h2 modifies the loss, we need to know [A]
        - How a change in layer h2 affects layer h3
        - How a change in layer h3 changes the model output
        - How a change in the model output changes the loss
    - As another example, to calculate how a small change in a weight/bias in h1 modifies the loss, we need to know [B]
        - How a change in layer h1 affects layer h2
        - How a change in layer h2 affects layer h3
        - How a change in layer h3 changes the model output
        - How a change in the model output changes the loss
    - In turn, we can find that there are many quantities in A that we resued in B (when computing derivatives w.r.t parameters in the earlier layer)
    - It follows that we can calculate many quantities for a layer and reuse them in the earlier layers
    - Proceeding backward through the network in this way to compute the derivatives is known as the **backward pass**

### Backpropagation Algorithm for 3-Layers MLP
- Forward pass: we work through all below calculations and store all the intermediate quantities
$$
\begin{align*}
f_0 &= \beta_0 + \Omega_0 x_i \\
h_1 &= a[f_0] \\
f_1 &= \beta_1 + \Omega_1 h_1 \\
h_2 &= a[f_1] \\
f_2 &= \beta_2 + \Omega_2 h_2 \\
h_3 &= a[f_2] \\
f_3 &= \beta_3 + \Omega_3 h_3 \\
\ell_i &= l[f_3, y_i]
\end{align*}
$$
- Backward pass 1:
    - First compute $\frac{\partial \ell_i}{\partial f_k}$, that is how the loss changes when the pre-activations $f_0$,$f_1$,$f_2$ change 
    - Applying the chain rule, the expression for the derivative of the loss $ℓ_i$ (scalar) with respect to each pre-activation (vector) is:
    $$
    \frac{\partial \ell_i}{\partial f_2}=
    \frac{\partial h_3}{\partial f_2}
    \frac{\partial f_3}{\partial h_3}
    \frac{\partial \ell_i}{\partial f_3}
    $$
    $$
    \frac{\partial \ell_i}{\partial f_1}=
    \frac{\partial h_2}{\partial f_1}
    \frac{\partial f_2}{\partial h_2}
    (\frac{\partial h_3}{\partial f_2}
    \frac{\partial f_3}{\partial h_3}
    \frac{\partial \ell_i}{\partial f_3})
    $$
    $$
    \frac{\partial \ell_i}{\partial f_0}=
    \frac{\partial h_1}{\partial f_0}
    \frac{\partial f_1}{\partial h_1}
    (\frac{\partial h_2}{\partial f_1}
    \frac{\partial f_2}{\partial h_2}
    \frac{\partial h_3}{\partial f_2}
    \frac{\partial f_3}{\partial h_3}
    \frac{\partial \ell_i}{\partial f_3})
    $$
    - Note that in each case, the term in brackets was computed in the previous step
    - By working backward through the network, we can reuse the previous computations
    - Each time, we only need to compute $\frac{\partial h_k}{\partial f_{k-1}}$ and $\frac{\partial f_k}{\partial h_k}$:
    $$\frac{\partial f_k}{\partial h_k} = \frac{\partial}{\partial h_k}\left( \beta_k + \Omega_k h_k \right) = \Omega_k^{T}$$
    $$\frac{\partial h_k}{\partial f_{k-1}} = \text{a diagonal matrix with diagonal element $j = 1$ if $f_{k-1, j} > 0$, 0 otherwise}$$
    - Rather than define the matrix $\frac{\partial h_k}{\partial f_{k-1}}$, we usually extract the diagonal terms as a vector $I[f_{k-1} > 0]$ and conduct pointwise multiply later on, which is more efficient
    - Now we can write a complete formula for $\frac{\partial \ell_i}{\partial f_{k-1}}$, which is:
    $$
    \frac{\partial \ell_i}{\partial f_{k-1}} =  \frac{\partial h_k}{\partial f_{k-1}}\frac{\partial f_k}{\partial h_k}\frac{\partial \ell_i}{\partial f_{k}} = \mathbb{I}[f_{k-1} > 0]\;\odot\; \left(\Omega_k^{T}\frac{\partial \ell_i}{\partial f_k}\right)
    $$
    - This formula applies for $k \in \{ K, K-1, ..., 1\}$, and notes that $⊙$ denotes pointwise multiplication
- Backward pass 2:
    - As we know how to compute $\frac{\partial \ell_i}{\partial f_k}$, we can focus on calculating the derivatives of the loss with respect to the weights and biases
    $$
    \frac{\partial \ell_i}{\partial \beta_k}=
    \frac{\partial f_k}{\partial \beta_k}
    \frac{\partial \ell_i}{\partial f_k}=
    \frac{\partial}{\partial \beta_k}
    \left( \beta_k + \Omega_k h_k \right)
    \frac{\partial \ell_i}{\partial f_k} =
    \frac{\partial \ell_i}{\partial f_k}
    $$
    $$
    \frac{\partial \ell_i}{\partial \Omega_k}=
    \frac{\partial f_k}{\partial \Omega_k}
    \frac{\partial \ell_i}{\partial f_k}=
    \frac{\partial}{\partial \Omega_k}
    \left( \beta_k + \Omega_k h_k \right)
    \frac{\partial \ell_i}{\partial f_k}=
    \frac{\partial \ell_i}{\partial f_k} h_k^{T}
    $$
    - Note that we compute $h_k$ at the forward pass
    - We calculate these derivatives for every training example in the batch and sum them together to retrieve the gradient for the SGD update
