# Regularization

## Generalization and Inductive Bias

### Cause of Poor Test Performance
- In machine learning, it is common to find performance gap between the training and test data
- Possible reasons for this generalization gap:
    - Over-fitting: the model describes statistical peculiarities of the training data that are not representative of the true mapping from input to output
        - e.g. the model fits training data perfectly
    - Poor Inductive Bias: the model is unconstrained in areas with no training examples (between data points), leading to suboptimal predictions
        - e.g. the model oscillates frequently between training data points

### Regulariation
- Generalization: performance of an estimated model on unseen (test) data
- Inductive bias: 
    - The tendency of a model to prioritiz solution with some properties between data points (e.g. smoothness)
    - Determines model behavior in underdetermined regimes and hence generalization
- Regulariation: techniques for imposing inductive bias / reduce over-fitting in model training and thus reduce the generalization gap between training and test performance

### Types of Regularization
- Explicit Regularization: adding explicit terms to the loss function that favor certain parameter choices
- Implicit Regularization: although not adding explicit penalty terms, the training algorithms may achieve results equivalent to adding penalty terms
    - e.g. the stochastic gradient descent algorithm is equivalent to using gradient descent with penalty terms added to the loss function
- Heuristics to Improve Performance: a set of heuristic methods that improve test performance

## Explicit Regularization

### Implicit Regularization and Probabilistic Interpretation
- Set Up:
    - Consider fitting a model $f[x, ϕ]$ with parameters $ϕ$ using a training set $\{x_i, y_i\}$ of input/output pairs
    - We seek the parameters $\hat ϕ$ that minimize the cross-entrophy loss (or negative log-likelihood):
    $$\begin{align*} \hat{\phi} = \arg\min_{\phi} \sum_{i=1}^I l_i[x_i,y_i] \end{align*}$$
    - To bias this minimization toward certain solutions (parameters) that we favour, we include an additional penalty term:
    $$\begin{align*} \hat{\phi} = \arg\min_{\phi} \sum_{i=1}^I l_i[x_i,y_i] + \lambda g[\phi] \end{align*}$$
    - Where $g[ϕ]$ is a function that returns a scalar which takes larger values when the parameters are less preferred
    - The term $λ$ is a positive scalar that controls the relative contribution of the original loss function and the penalty term
- Probabilistic interpretation:
    - Maximum a posteriori (MAP): 
        - Suppose that prior probability distribution $Pr(ϕ)$ represents our knowledge about the parameter distribution
        - The posteriori probability of $ϕ$ realized given that we observe a set of $x_i$ and $y_i$ pairs can be written as:
        $$ 
        \begin{align*} P(\phi | D) & = \frac{P(D|\phi) P(\phi)}{P(D)}  \\
        & = \frac{P(\{y_i\}|\{x_i\}, \phi) P(\{x_i\}|\phi) P(\phi)}{P(\{y_i\}|\{x_i\})P(\{x_i\})} \\
        & = \frac{P(\{y_i\}|\{x_i\}, \phi) P(\{x_i\}) P(\phi)}{P(\{y_i\}|\{x_i\})P(\{x_i\})} \\
        & = \frac{P(\{y_i\}|\{x_i\}, \phi) P(\phi)}{P(\{y_i\}|\{x_i\})} \\
        & \propto \prod_{i=1}^I q(y_i|x_i, \phi) P(\phi)
        \end{align*}$$
        - Therefore, the maximization problem that solves the $\hat ϕ$ with highest posteriori probaility of relaization can be written as:
        $$
        \begin{align*} \hat{\phi} &= \arg\max_{\phi} \prod_{i=1}^I q(y_i|\theta_i=f[x_i, \phi]) P(\phi) \\ 
        &= \arg\max_{\phi} \log \prod_{i=1}^I q(y_i|\theta_i=f[x_i, \phi]) P(\phi)\\ 
        &= \arg\min_{\phi} - \log P(\phi) -\sum_{i=1}^I \log q(y_i|\theta_i=f[x_i, \phi]) \\ 
        &= \arg\min_{\phi} \lambda g[\phi] + \sum_{i=1}^I l_i[x_i,y_i] 
        \end{align*}
        $$
        - Hence, explicit regularization is equivalent to find a parameter $ϕ$ with maximized posteriori probability of relaization, given a prior distribution of the parameter $P(\phi)$
        - The assumption on prior distribution of the parameter $P(\phi)$ will determine the form of the penalty term $\lambda g[\phi] = - \log P(\phi)$

### L2 Regularization
- The most commonly used regularization (penalty) term is the L2 norm
    - It penalizes the sum of the squares of the parameter values:
    $$\begin{align*} g[\phi] = \sum_j \phi_j^2 \end{align*}$$
    - Thus the minimization problem that the parameter solves is:
    $$\begin{align*} \arg\min_{\phi} \sum_{i=1}^I l_i[x_i,y_i] + \lambda \sum_j \phi_j^2 \end{align*}$$
    - This is also referred as ridge regression
- Porperties:
    - The effect is to encourage smaller parameter, so the output function (of complex model, e.g. neural network) will be smoother
    - Higher $\lambda$, more smooth the output function is
    - Address over-fitting:
        - If the model is overfitting, adding the regularization means that the model must trade off perfect-fitting of the training data against the desire to be smooth
        - The error due to variance reduces (the model no longer fit every data point) at the cost of increased bias (the model can only describe smooth functions)
    - Address poor inductive bias:
        - When a network is over-parameterized, the extra model capacity describes areas with no training data
        - Thus, the regularization term will favor functions that smoothly interpolate between the data points

### L1 Regularization
- The penalty term penalizes the sum of absolute value of the parameters: $$\begin{align*} g[\phi] = \sum_j |\phi_j| \end{align*}$$
- Different from L2 Regularization, L1 regularization will bias some paremters toward 0 (thus conduct feature selection)

## Implicit Regularization

### Gradient Descent
- Consider a continuous version of gradient descent where the step size is infinitesimal
- The change in parameters $ϕ$ will be governed by the differential equation: $$\frac{dϕ}{dt} = −\frac{∂L}{∂ϕ}$$
- Gradient descent approximates this process with a series of discrete steps of size $α$: $$ϕ_{t + 1} = ϕ_{t} − \alpha \frac{∂L[ϕ_t]}{∂ϕ}$$
- The discretization causes a deviation from the continuous path
- However, we can derive a modified loss function $\tilde{L}$ for the continuous case that arrives at the same place as the discretized version on the original loss $L$
- It can be shown that: $$\tilde{L}_{GD}[\phi] = L[\phi] + \frac{\alpha}{4} \left\| \frac{\partial L}{\partial \phi} \right\|^2 $$
- In other word, the discrete gradient descent follows a continous path that minimizes a loss function that penalizes (regularizes) large gradient
- When $\alpha$, the learning ate is higher, the penalty on large graident is higher, which mights explain why large $\alpha$ leads to model that generalizes better

### Stochastic Gradient Descent
- We can also formulate a modified loss function such that the continuous version of gradient descent reaches the same local minimum as the SGD
- Discrete stochastic gradient descent on $L$ arrives at same place as continuous descent $\frac{d\phi}{dt} = - \frac{\partial\tilde{L}}{\partial\phi}$ with :
$$\begin{align*} \tilde{L}[\phi] = L[\phi] + \frac{\alpha}{4} ||\frac{\partial L}{\partial\phi}||^2 + \frac{\alpha}{4B}\sum_{b=1}^B || \frac{\partial L_b}{\partial\phi} - \frac{\partial L}{\partial\phi} ||^2 \end{align*}$$
- $L_b$ is the mean of the individual losses for the $b$ th of the $B$ batches in an epoch
- $L$ is the mean of the I individual losses in the full traning set
- Therefore, comparing to GD, SGD adds an additional regularization term that penalizes the variance of the gradients of the batch losses
- That is, SGD favors updates where the gradients are stable (all gradient of batch loss close to gradient of training set loss)
- When updating, it is willing to trade some reduction in loss for gradient stability across batches
- Hence, SGD generalizes for two potential reasons:
    - Inherent randomness allows the algorithm to reach different parts of the loss function (escape from local minimum)
    - Implicit regularization that encourages solutions where all the data fits evenly well rather than solutions where some of the data fit extremely well and other data less well 

##  Heuristics to Improve Performance

### Early Stopping
- Refers to stopping the training procedure before it has fully converged
- This can reduce overfitting if the model has already captured the coarse shape of the underlying function but has not yet overfit to the noise
- In practice, halt training when validation loss starts rising
- Avoids overfitting by limiting effective model complexity with out changing architecture

### Ensembling
- Refers to build several models and average their predictions
- The models can be combined by:
    - Taking the mean of the outputs (for regression problems) 
    - Taking the median of the outputs (for regression problems) 
    - Taking the mean of the pre-softmax activations (for classification problems)
    - Taking the most frequent predicted class (for classification problems)
- The assumption is that model errors are independent and thus averaging models reduce the variance of the model

### Drop Out (for Neural Network)
- Dropout clamps a random subset (typically 50%) of hidden units to zero at each iteration of SGD
- This makes the neural network less dependent on any given hidden unit
- It encourages the model weights to have smaller magnitudes so the change in function due to presence/absence of specific hidden unit is reduced
- It is similar as explicit regularization: again, if weights are small, the output function is more likley to be smooth (less likely to overfit / oscillates between data points)

### Applying Noise
- Refers to add noise to inputs, model weights, or labels during training
- Acts as a smoothing penalty and improves robustness: this smooths out the output function
- For regression problems, it can be shown to be equivalent to adding a regularizing term that penalizes the derivatives of the model output with respect to its input
- Adversarial training is a variant of this technique: the algorithm actively searches for noise to the input that cause large changes to the output

### Bayesian Inference
- Instead of enforcing point estimates of the parameters, Bayesian inference approach estimates a posterior distribution over parameters: 
$$P(\phi|\{x_i,y_i\}) \propto \prod_{i=1}^I q(y_i|x_i,\phi)P(\phi)$$
- The prediction y for new input x is an infinite weighted sum of the predictions for each parameter set, where the weights are the associated probabilities: 
$$\Pr(y \mid \mathbf{x}, \{x_i, y_i\}) = \int \Pr(y \mid \mathbf{x}, \phi)\Pr(\phi \mid \{x_i, y_i\})\, d\phi$$
- The Bayesian approach is elegant and can provide more robust predictions than point estimates of the parameters (e.g. maximum likelihood)
- In general, Bayesian approach will make output function also smoother
- However, these approaches typically add considerable complexity to learning and inference

### Normalization 
- Refers to re-center and re-scale activations inside a neural network layer during training
- Batch normalization: normalize an activation (one element of the vector output of a layer) by mean / variance of the batch
- Layer normalization: normalize an activation by mean / variance across activations of the same layer
- Group normalization: normalize over groups (useful when batch size small)

### Transfer Learning
- Transfer learning: (useful when data limited)
    - The network is pre-trained to perform a related secondary task for which data are more plentiful
    - The resulting model is then adapted to the original task
    - This is typically done by removing the last layer and adding one or more layers that produce a suitable output
    - Or through finetuning the whole model using data for the original task (update all paremeters of the trained model)
    - Intuition: 
        - The network will build a good internal representation of the data from the secondary task
        - Which can subsequently be exploited for the original task
- Multi-task learning:
    - The network is trained to solve several problems concurrently
    - For example, the network might take an image and simultaneously learn to segment the scene and estimate the pixel-wise depth
    - Both tasks require some understanding of the image and, when learned simultaneously, the model performance for each may improve

### Self-supervised Learning
- For transfer learning, if we do not have plentiful data for secondary task, we may consider about self-supervised learning as pre-training
- Generative self-supervised learning: part of each data example is masked, and the secondary task is to predict the missing part
- Contrastive self-supervised learning: pairs of examples with commonalities are compared to unrelated pairs
    - Text example: to determine whether two sentences followed one another in the original document
    - Image example: identify whether a pair of images are transformed versions of one another or are unconnected
    - The "label" of these tasks are created naturally (e.g. transformation of image is imposed by us / the raw text data is at document level)

### Data Augmentation
- The principle of data augmentation is to expand data set and teach the model to be indifferent to irrelevant data transformations
- We can often transform each input data example in such a way that the label stays the same 
- For example, we might aim to determine if there is a bird in an image
- Then we could rotate, flip, blur, or manipulate the color balance of the image, and the label “bird” remains valid

## Ensemble and Tree-based Models

### Classification and Regression Trees
- A regression or classfication tree partitions the entire input space into several regions using axis parallel splits
- The prediction of output to X is the mean (for regression) or mode (for classfication) of outputs associated with features in the partition that X belong to
- At each node, regression / classification tree chooses feature + threshold to minimize impurity at that node:
    - Impurity for regression task: MSE
    - Impurity for classification task: entropy or Gini index (penalizes diverse label categories in the nodes / if one split creates two leaf nodes that each contains only one label, perfect purity with entrophy / Gini = 0)

### Bagging and Random Forests
- Trees are interpretable, but usually with high variance (sensitive to training data)
- There are two solutions that targets this limitation: bagging and random forest
- Bagging:
    - Fit many trees in parallel on bootstrap samples of the training data and average their predictions to form the output (bootstrap aggregating)
    - Averaging (or voting in case of classfication) reduces variance of the model
    - Bootstrap sample: sampling from the original training data with replacement
- Random forests:
    - Although bagging use random sample to train each model, if there exist significant feature, models still highly correalted, averaging does not reduce variance by much
    - To tackle this, random forest not only use bootstrap sample, but also at each split use a random subset of features
    - This further decorrelates trees (and model error), thus decreasing model variance

### Boosting and Gradient Boosting
- Boosting:
    - A sequential of models (trees): each new model focuses on errors (residuals, applicable in regression) of the previous ones to iteratively improve weak learners
    - Add predictions of models up to form the ensemble model output
- Gradient Boosting:
    - Sequentially fit shallow trees to the negative gradient of the loss of previous one (also called pseudo-residuals, works for any differentiable loss, so applicable to both regression and classification) 
    - Each tree adds a small pointing in the direction that most reduces prediction error
    - The output is again the sum of predictions over trees
- XGBoost: 
    - Effcient implementation of Gradient Boosting (state-of-the-art for tabular data)
    - L2 regularization on leaf weights, second-order (Newton) splits, column subsampling, and efficient handling of sparse data

## Model Interpretability

### Shapley Values
- Shapley value: measures a feature's average marginal contribution to the prediction, averaged over all possible subsets of other features
- The measure is based on the pinciples:
    - Fair attribution (cooperative game theory): a unique rule assigning feature contribution that satisfying 
        - Efficiency: prediction = baseline + sum of additive feature contributions
        - Symmetry: if two features contribute equally in every possible subsets of other features, then they must receive equal Shapley values
        - Consistency: if a model changes so a feature contributes more, its Shapley value does not decrease
    - Features = players
    - Contribution = each player’s share of payoff
    - Baseline = payoff when no players participate
- SHAP: practical method to approximate Shapley values
