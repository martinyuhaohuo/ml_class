# Representation Learning

## Convolutional Neural Networks

### Reasons for CNN
- Images have three properties that suggest the need for specialized model architecture:
    1. Images are high-dimensional: this poses practical problems in terms of memory and computation if applying full connected MLP
    2. Nearby image pixels are statistically related:  but MLP have no notion of “nearby” and treat the relationship between every input pixel equally
    3. Third, the interpretation of an image is stable under geometric transformations (e.g. shift leftward a bit): but MLP cannot guarantee this
- Convolutional layers process each local image region independently, using parameters shared across the whole image
- A network predominantly consisting of convolutional layers is known as a **convolutional neural network** (CNN)
- Advantage of CNN:
    - They use fewer parameters than fully connected MLP
    - They exploit the spatial relationships between nearby pixels
    - They don’t have to re-learn the interpretation of the pixels at every position
    - They usually achieve **Invariance** for classification task, **Equivariance** for segmentation task

### Invariance and Equivariance
- Invariant: a function $f[x]$ of an image x is invariant to a transformation $t[x]$ if $f[t[x]]= f[x]$
    - The network $f[x]$ for image classification should be invariant to geometric transformations of the image
    - The network $f[x]$ should identify an image as containing the same object, even if it has been translated, rotated, flipped, or warped
- Equivariant: a function f[x] of an image x is equivariant to a transformation $t[x]$ if $f[t[x]]= t[f[x]]$
    - The network $f[x]$ for per-pixel image segmentation should be equivariant to transformations
    - If the image is translated, rotated, or flipped, the network $f[x]$ should return a segmentation that has been transformed in the same way

### 1D Convolution Operation
- Convolutional operation (convolving):
    - In 1D, a convolution transforms an input vector $x$ into an output vector $z$ so that each output $z_i$ is a weighted sum of nearby inputs
- Kernel size:
    - The size of the region over which inputs are combined is termed the **kernel size**
    - For a kernel size of three, we have the convolution layer as: $z_i = ω_1x_{i−1} + ω_2x_i + ω_3x_{i+1}$
    - The kernel size typically remains an odd number so that it can be centered around the current position
- Kernel:
    - Where $ω = [ω_1, ω_2, ω_3]^T$ is called **kernel**, note that the convolution opeation is equivariant with respect to translation
- Stride:
    - When we evaluate the output at every position, we term this a stride of one
    - However, it is also possible to shift the kernel by a stride greater than one
    - If we have a stride of two, we create roughly half the number of outputs
- Dilation rate:
    - Dilated convolution have kernel values that are interspersed with zeros
    - For example, we can turn a kernel of size five into a dilated kernel of size three by setting the second and fourth elements to zero
    - The number of zeros we intersperse between the weights determines the dilation rate
- Padding:
    - Each output of the convolution layer is computed by taking a weighted sum of the previous, current, and subsequent positions in the input
    - This begs the question of how to deal with the first output (no previous input) and the final output (no subsequent input)
    - There are two common approaches:
        - **Zero-padding** assumes the input is zero outside its valid range
        - **Valid convolutions** is to discard the output positions where the kernel exceeds the range of input positions
- Convolutional layer:
    - Convolutional layer computes its output by convolving the input, adding a bias $β$, and passing each result through an activation function $a[•]$
    - With kernel size 3, stride 1, and dilation rate 1, the i-th hidden unit $h_i$ would be computed as:
    $$
    h_i = a [β + ω_1x_{i−1} + ω_2x_i + ω_3x_{i+1}]
    $$
    - Where the bias $β$ and kernel weights $ω_1, ω_2, ω_3$ are trainable parameters
    - This is a special case of a hidden unit in a fully connected layer (if there are D inputs and D hidden units) that computes the ith hidden unit as
    $$
    h_i = a [β + \sum_{j = 1}^{D} ω_{ij}x_{j}]
    $$
- Channels:
    - If we only apply a single convolution, information will likely be lost
    - Hence, it is usual to compute several convolutions in parallel
    - Each convolution produces a new set of hidden variables, termed **channel**
    - Note that we can continue to apply convolution layers on channels of input data

### Convolutional networks and receptive fields
- Convolutional networks comprise a sequence of convolutional layers
- The receptive field of a hidden unit in the network is the region of the original input that feeds into it
-  Consider a convolutional network where each convolutional layer has kernel size three
- The hidden units in the first layer take a weighted sum of the three closest inputs, so have receptive fields of size three
- The units in the second layer take a weighted sum of the three closest positions in the first layer, which are themselves weighted sums of three inputs
- Hence, the hidden units in the second layer have a receptive field of size five
- In this way, the receptive field of units in successive layers increases, and information from across the input is gradually integrated
